# The Data

```python
import pandas as pd
```

```python
df = pd.read_csv("Boston House Prices.csv")
```

```python
import statsmodels.api as sm
```

```python
from sklearn import linear_model
```

# Linear Regression with Statsmodels

```python
#dependent variable / inpendent variable
y = df['Value'] # value is dependent variable
x = df['Rooms'] # x = independent
```

```python
df
```

### Exploring the dataset


```python
#create plot()  kind, independent, dependent
df.plot(kind = 'scatter', x = 'Rooms', y = 'Value')
```

### Making the regression: Adding a constant and fitting the model


Statmodels doesn't add a constant by default, so we have to do it on our own


```python
# adding a constant for independent varable
x = sm.add_constant(x)
```

```python
# fitting the model dependent, independent variable
lm = sm.OLS(y,x).fit()
```

```python
#predict
lm.predict(x)
```

### The Regression Table


```python
lm.summary()
```

### Linear Regression Equation


```python
# Rooms coef: 9.1021 = a
# Constant coef: - 34.6706 = b

# Linear equation: 𝑦 = 𝑎𝑥 + 𝑏 x = independent variable
equation = 9.1021*x['Rooms'] - 34.6706
```

### Plotting The Regression Line


```python
# plotting the data points
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
plt.figure(tight_layout=True)
#ploting data point
sns.scatterplot(x = 'Rooms', y = 'Value', data = df)
# sns.line plot the line
sns.lineplot(x = 'Rooms', y = equation, data = df, color = 'green')
#axes
plt.xlabel(0);
plt.ylabel(0);
plt.savefig('linear_regression')
plt.show()
```

# Multiple linear regression

## Define the dependent and independent variable

```python
x = df[['Rooms', 'Distance']] # independent variable
y = df['Value']
```

## Making the regression: Adding a constant and fitting the model

# Linear Regression with Scikit-learn

we dont need add constant independent variable for scikit learn linear

## Define the dependent and independent variable

```python
x = df[['Rooms', 'Distance']]
y = df['Value']
```

## Fitting the Model

```python
lm = linear_model.LinearRegression()
lm.fit(x,y)
```

## Predicting values

```python
lm.predict(x)
```

## Regression Table

```python
lm.score(x,y)
```

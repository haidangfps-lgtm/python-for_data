```python
import pandas as pd
```

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv')
```

```python
# select columns
df_movies = df_movies[['imdb_title_id', 'title', 'year',
                       'genre', 'country']]

df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

# merge()


```python
df1 = pd.DataFrame({'id': ['A', 'B', 'C', 'D'] , 'age': [30, 23, 25, 22]})
```

```python
df2 = pd.DataFrame({'id': ['C', 'D', 'E', 'F'] , 'job': ['Doctor', 'Statistician','Accountant', 'Developer']})
```

```python
df1.merge(df2, on = 'id', how='inner')
```

```python
pd.merge(df1, df2, on = 'id', how='inner')
```

## Inner join


### Exercise


```python
# merge df_movies and df_ratings (inner join) 2ways
df_movies.merge(df_ratings, on = 'imdb_title_id', how='inner')
```

```python
pd.merge(df_ratings, df_movies, on = 'imdb_title_id', how='inner')
```

## Outer join (Full join)


### Exercise


```python
# merge df_movies and df_ratings (outer join)
pd.merge(df_ratings, df_movies, on = 'imdb_title_id', how='outer')
```

## Exclusive Outer join (Exclusive Full join)


```python
# indicator=True
pd.merge(df1, df2, on = 'id', how='outer', indicator=True)
```

```python
# query
df = pd.merge(df1, df2, on = 'id', how='outer', indicator=True).query("_merge == 'left_only' or _merge == 'right_only'")
```

```python
# shape
df.shape
```

```python
df1.shape
```

```python
df2.shape
```

##  Left join


```python
df1.merge(df2, on = 'id', how = 'left', indicator=True)
```

## Exclusive Left join


```python
# indicator=True
pd.merge(df1, df2, on = 'id', how='left',indicator=True)
```

```python
# query
pd.merge(df1, df2, on = 'id', how='left',indicator=True).query("_merge == 'left_only'")
```

### Exercise


```python
# make a copy of the df_movies dataframe
df = df_movies.copy()
```

```python
# set the first 1000 values of 'imdb_title_id' column as 'tt1234567890'
for i in df.index:
    if i < 1000:
        df.loc[i, 'imdb_title_id'] = 'tt1234567890'
```
